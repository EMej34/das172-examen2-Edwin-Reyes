```python
# ==============================
# Módulo 1: Validación Dimensional
# ==============================
def validar_dimensiones(cargas, capacidades):
    if len(cargas) < 2 or len(capacidades) < 2:
        return False
    if len(cargas) != len(capacidades):
        return False
    for fila_c, fila_cap in zip(cargas, capacidades):
        if len(fila_c) != len(fila_cap):
            return False
        for peso, cap in zip(fila_c, fila_cap):
            if peso < 0 or cap <= 0:
                return False
    return True


# ==============================
# Módulo 2: Ocupación y Sobrecarga
# ==============================
def calcular_ocupacion(cargas, capacidades):
    N, M = len(cargas), len(cargas[0])
    matriz_ocupacion = [[0]*M for _ in range(N)]
    sobrecargadas = []
    for i in range(N):
        for j in range(M):
            porcentaje = (cargas[i][j] / capacidades[i][j]) * 100
            matriz_ocupacion[i][j] = porcentaje
            if porcentaje > 100:
                sobrecargadas.append((i, j))
    return matriz_ocupacion, sobrecargadas


# ==============================
# Módulo 3: Balance y Simetría
# ==============================
def evaluar_balance(cargas, tolerancia):
    N, M = len(cargas), len(cargas[0])
    pesos_fila = [sum(fila) for fila in cargas]

    mitad = M // 2
    if M % 2 == 0:
        izquierda = sum(cargas[i][j] for i in range(N) for j in range(mitad))
        derecha = sum(cargas[i][j] for i in range(N) for j in range(mitad, M))
    else:
        izquierda = sum(cargas[i][j] for i in range(N) for j in range(mitad))
        derecha = sum(cargas[i][j] for i in range(N) for j in range(mitad+1, M))

    desbalance = abs(izquierda - derecha)
    balance_ok = desbalance <= tolerancia

    return pesos_fila, desbalance, balance_ok


# ==============================
# Módulo 4: Submatriz Crítica
# ==============================
def submatriz_critica(matriz_ocupacion, k, p):
    N, M = len(matriz_ocupacion), len(matriz_ocupacion[0])
    mejor_promedio = -1
    mejor_submatriz = None

    for i in range(N - k + 1):
        for j in range(M - p + 1):
            sub = [fila[j:j+p] for fila in matriz_ocupacion[i:i+k]]
            promedio = sum(sum(fila) for fila in sub) / (k*p)
            if promedio > mejor_promedio:
                mejor_promedio = promedio
                mejor_submatriz = sub
    return mejor_submatriz


# ==============================
# Script Principal de Ejecución
# ==============================
def ejecutar_prueba(cargas, capacidades, tolerancia=300):
    print("=== VALIDACIÓN DE MATRICES ===")
    print("Resultado:", validar_dimensiones(cargas, capacidades))

    print("\n=== CÁLCULO DE OCUPACIÓN ===")
    ocupacion, sobrecargadas = calcular_ocupacion(cargas, capacidades)
    for fila in ocupacion:
        print(["{:.2f}%".format(val) for val in fila])
    print("Celdas sobrecargadas:", sobrecargadas)

    print("\n=== EVALUACIÓN DE BALANCE ===")
    pesos_fila, desbalance, balance_ok = evaluar_balance(cargas, tolerancia)
    for i, peso in enumerate(pesos_fila, start=1):
        print(f"Fila {i}: {peso:.2f} kg")
    print(f"Desbalance lateral: {desbalance:.2f} kg")
    print("Balance aprobado:", "Sí" if balance_ok else "No")

    print("\n=== SUBMATRIZ CRÍTICA ===")
    sub = submatriz_critica(ocupacion, 2, 2)
    for fila in sub:
        print(["{:.2f}%".format(val) for val in fila])


# ==============================
# Pruebas Unitarias con resultados separados
# ==============================
def pruebas_unitarias():
    print("\n=== PRUEBAS UNITARIAS ===")

    try:
        print("Prueba validación correcta:", validar_dimensiones([[100,200],[300,400]], [[200,200],[400,400]]))
        print("Prueba dimensiones incorrectas:", validar_dimensiones([[100,200],[300,400]], [[200,200]]))
        print("Prueba valores inválidos:", validar_dimensiones([[-10,200],[300,400]], [[200,200],[400,400]]))

        ocupacion, sobrecargadas = calcular_ocupacion([[100,200],[300,400]], [[200,200],[400,400]])
        print("Prueba ocupación normal:", ocupacion, "Sobrecargadas:", sobrecargadas)

        _, sobrecargadas = calcular_ocupacion([[300,200],[500,400]], [[200,200],[400,400]])
        print("Prueba ocupación con sobrecarga:", sobrecargadas)

        _, desbalance, balance_ok = evaluar_balance([[100,100],[200,200]], tolerancia=50)
        print("Prueba balance equilibrado: Desbalance =", desbalance, "Balance OK =", balance_ok)

        _, desbalance, balance_ok = evaluar_balance([[500,100],[200,50]], tolerancia=100)
        print("Prueba balance desequilibrado: Desbalance =", desbalance, "Balance OK =", balance_ok)

        matriz = [[50,60,70],[80,90,100],[40,30,20]]
        sub = submatriz_critica(matriz, 2, 2)
        print("Prueba submatriz crítica:", sub)

    except AssertionError:
        print("❌ Alguna prueba falló")
    else:
        print("✅ Todas las pruebas ejecutadas correctamente")


# ==============================
# Ejecución principal
# ==============================
if __name__ == "__main__":
    cargas = [
        [500, 600, 700, 800],
        [400, 300, 200, 100],
        [250, 350, 450, 550],
        [600, 500, 400, 300],
        [150, 250, 350, 450]
    ]
    capacidades = [
        [600, 600, 800, 900],
        [500, 400, 300, 200],
        [300, 400, 500, 600],
        [700, 600, 500, 400],
        [200, 300, 400, 500]
    ]

    ejecutar_prueba(cargas, capacidades)
    pruebas_unitarias()
```
